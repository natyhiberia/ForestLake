Convertir numericos en Silver

```python
from pyspark.sql.functions import col, trim, regexp_replace, when

# 1. Leer Bronze otra vez
df_raw = spark.table("forestlake.bronze.inventario_forestal_raw")

# 2. Nombres correctos
clean_columns = [
    "ID", "Familia", "Nombre científico", "Nombre común",
    "CoordX_WGS84", "CoordY_WGS84",
    "CAP1_cm", "CAP2_cm", "CAP3_cm", "CAP4_cm", "CAP5_cm", "CAP6_cm",
    "DAP_Total", "HT_m", "HC_m", "VOL_Total_m3", "Area_Basal_m2",
    "E_Fitosanitario", "Flor", "Fruto", "Luz", "Poste", "Lineas", "Observaciones"
]

# 3. Quitar encabezado ANTES de renombrar
df_no_header = df_raw.filter(col("_c0") != "ID")

# 4. Renombrar
df_silver = df_no_header.toDF(*clean_columns)

# 5. Nombres técnicos para Delta
rename_map = {
    "ID": "id",
    "Familia": "familia",
    "Nombre científico": "nombre_cientifico",
    "Nombre común": "nombre_comun",
    "CoordX_WGS84": "coordx_wgs84",
    "CoordY_WGS84": "coordy_wgs84",
    "CAP1_cm": "cap1_cm",
    "CAP2_cm": "cap2_cm",
    "CAP3_cm": "cap3_cm",
    "CAP4_cm": "cap4_cm",
    "CAP5_cm": "cap5_cm",
    "CAP6_cm": "cap6_cm",
    "DAP_Total": "dap_total",
    "HT_m": "ht_m",
    "HC_m": "hc_m",
    "VOL_Total_m3": "vol_total_m3",
    "Area_Basal_m2": "area_basal_m2",
    "E_Fitosanitario": "estado_fitosanitario",
    "Flor": "flor",
    "Fruto": "fruto",
    "Luz": "luz",
    "Poste": "poste",
    "Lineas": "lineas",
    "Observaciones": "observaciones"
}

for old, new in rename_map.items():
    df_silver = df_silver.withColumnRenamed(old, new)

# 6. Limpiar espacios en textos
for c in df_silver.columns:
    df_silver = df_silver.withColumn(c, trim(col(c)))

# 7. Convertir números usando try_cast para evitar errores
from pyspark.sql.functions import expr

numeric_cols = [
    "coordx_wgs84", "coordy_wgs84",
    "cap1_cm", "cap2_cm", "cap3_cm", "cap4_cm", "cap5_cm", "cap6_cm",
    "dap_total", "ht_m", "hc_m", "vol_total_m3", "area_basal_m2"
]

for c in numeric_cols:
    df_silver = df_silver.withColumn(c, expr(f"try_cast({c} as double)"))

for c in numeric_cols:
    df_silver = df_silver.withColumn(c, col(c).cast("double"))

# 8. Filtrar registros válidos
df_silver_final = df_silver.filter(
    col("id").isNotNull() &
    col("coordx_wgs84").isNotNull() &
    col("coordy_wgs84").isNotNull()
)

display(df_silver_final)
```

Save in silver

```python
df_silver_final.write.mode("overwrite").saveAsTable(
    "forestlake.silver.inventario_forestal_limpio"
)
```

```python
# ENRIQUECIMIENTO SILVER


from pyspark.sql.functions import when, col

df_silver_enriched = (
    df_silver_final

    .withColumn(
        "riesgo_dap",
        when(col("dap_total") >= 80, "Alto")
        .when(col("dap_total") >= 40, "Medio")
        .otherwise("Bajo")
    )

    .withColumn(
        "riesgo_altura",
        when(col("ht_m") >= 15, "Alto")
        .when(col("ht_m") >= 8, "Medio")
        .otherwise("Bajo")
    )

    .withColumn(
        "interferencia_urbana",
        when(
            (col("poste").isNotNull()) |
            (col("lineas").isNotNull()) |
            (col("luz").isNotNull()),
            "Si"
        ).otherwise("No")
    )
)

display(df_silver_enriched)
```

Bajo: árboles menores de 8 m

```python
# Leer tabla silver

df_silver = spark.table("forestlake.silver.inventario_forestal_limpio")

display(df_silver)
```

```python
# Crear dataframe enriquecido

df_silver_enriched = df_silver
```

```python

# CATEGORÍA DE ALTURA


from pyspark.sql.functions import when, col

df_silver_enriched = df_silver_enriched.withColumn(
    "categoria_altura",
    when(col("ht_m") >= 15, "Alto")
    .when(col("ht_m") >= 8, "Medio")
    .when(col("ht_m").isNotNull(), "Bajo")
    .otherwise("Sin dato")
)

display(df_silver_enriched)
```

```python
df_silver_enriched = df_silver_enriched.withColumn(
    "rango_altura",
    when(col("ht_m") < 5, "0 - 4.9 m")
    .when((col("ht_m") >= 5) & (col("ht_m") < 10), "5 - 9.9 m")
    .when((col("ht_m") >= 10) & (col("ht_m") < 15), "10 - 14.9 m")
    .when((col("ht_m") >= 15) & (col("ht_m") < 20), "15 - 19.9 m")
    .otherwise("> 20 m")
)
```

CATEGORIA DE DAP

DAP alto: ≥ 80 cm
DAP medio: 40–79.9 cm
DAP bajo: < 40 cm

```python

#  CATEGORÍA DE DAP

df_silver_enriched = df_silver_enriched.withColumn(
    "categoria_dap",
    when(col("dap_total") >= 80, "DAP alto")
    .when(col("dap_total") >= 40, "DAP medio")
    .when(col("dap_total").isNotNull(), "DAP bajo")
    .otherwise("Sin dato")
)

display(df_silver_enriched)
```

Estado Fenologico: Si tiene presencia de Flor o Fruto

```python
# VARIABLES FENOLÓGICAS


df_silver_enriched = (
    df_silver_enriched
    .withColumn(
        "tiene_flor",
        when(col("flor") == "X", 1).otherwise(0)
    )
    .withColumn(
        "tiene_fruto",
        when(col("fruto") == "X", 1).otherwise(0)
    )
)

display(df_silver_enriched)
```

```python
from pyspark.sql.functions import col, when, trim, upper

df_silver_enriched = (
    df_silver_enriched
    .withColumn("luz_norm", upper(trim(col("luz"))))
    .withColumn("poste_norm", upper(trim(col("poste"))))
    .withColumn("lineas_norm", upper(trim(col("lineas"))))
    .withColumn(
        "interferencia_urbana",
        when(
            (col("luz_norm") == "X") |
            (col("poste_norm") == "X") |
            (col("lineas_norm") == "X"),
            "Si"
        ).otherwise("No")
    )
)

display(df_silver_enriched)
```

RIESGO URBANO: Interferencia de Lineas, Postes y Luz

```python
#RIESGO URBANO PRELIMINAR

df_silver_enriched = df_silver_enriched.withColumn(
    "riesgo_urbano",
    when(
        (col("categoria_altura") == "Alto") &
        (col("interferencia_urbana") == "Si"),
        "Alto"
    )
    .when(
        (col("categoria_altura") == "Medio") &
        (col("interferencia_urbana") == "Si"),
        "Medio"
    )
    .otherwise("Bajo")
)

display(df_silver_enriched)
```

FITOSANITARIO

```python
from pyspark.sql.functions import when

df_silver_enriched = df_silver_enriched.withColumn(
    "estado_fitosanitario_desc",
    when(col("estado_fitosanitario") == "B", "Bueno")
    .when(col("estado_fitosanitario") == "R", "Regular")
    .when(col("estado_fitosanitario") == "M", "Malo")
    .when(col("estado_fitosanitario") == "N", "Muerto")
    .otherwise("Sin dato")
)
```

# Schema Gold

```python
df_gold = spark.table("forestlake.silver.inventario_forestal_enriched")
```

Creacion del KPI General

```python
from pyspark.sql.functions import count, sum, when

kpi_general = df_gold.select(
    
    count("*").alias("total_arboles"),
    
    sum(
        when(col("interferencia_urbana") == "Si", 1)
        .otherwise(0)
    ).alias("arboles_con_interferencia"),
    
    sum("tiene_flor").alias("arboles_con_flor"),
    
    sum("tiene_fruto").alias("arboles_con_fruto"),
    
    sum(
        when(col("riesgo_urbano") == "Alto", 1)
        .otherwise(0)
    ).alias("arboles_riesgo_alto")
)

display(kpi_general)
```

Save KPI in Gold

```python
#  GUARDAR KPI GENERAL EN GOLD


kpi_general.write.mode("overwrite").saveAsTable(
    "forestlake.gold.kpi_general_inventario"
)
```

KPI ARBOLES POR FAMILIA

```python
# KPI ÁRBOLES POR FAMILIA


from pyspark.sql.functions import count

kpi_familia = (
    df_gold
    .groupBy("familia")
    .agg(count("*").alias("total_arboles"))
    .orderBy(col("total_arboles").desc())
)

display(kpi_familia)
```

```python
kpi_familia.write.mode("overwrite").saveAsTable(
    "forestlake.gold.kpi_arboles_por_familia"
)
```

KPI — Riesgo urbano por especie

```python
# KPI RIESGO URBANO POR ESPECIE


from pyspark.sql.functions import count

kpi_riesgo_especie = (
    df_gold
    .filter(col("riesgo_urbano") == "Alto")
    .groupBy("nombre_cientifico")
    .agg(
        count("*").alias("total_arboles_riesgo")
    )
    .orderBy(col("total_arboles_riesgo").desc())
)

display(kpi_riesgo_especie)
```

```python
kpi_riesgo_especie.write.mode("overwrite").saveAsTable(
    "forestlake.gold.kpi_riesgo_especie"
)
```

KPI Interferencias Urbanas

```python
#  KPI INTERFERENCIAS URBANAS

from pyspark.sql import Row

# Drop table if exists to avoid schema mismatch
spark.sql("DROP TABLE IF EXISTS forestlake.gold.kpi_interferencias_urbanas")

interferencias = [
    Row(tipo_interferencia="Luz",
        total_arboles=df_gold.filter(col("luz") == "X").count()),

    Row(tipo_interferencia="Poste",
        total_arboles=df_gold.filter(col("poste") == "X").count()),

    Row(tipo_interferencia="Lineas",
        total_arboles=df_gold.filter(col("lineas") == "X").count())
]

kpi_interferencias_urbanas = spark.createDataFrame(interferencias)

kpi_interferencias_urbanas.write.mode("overwrite").saveAsTable(
    "forestlake.gold.kpi_interferencias_urbanas"
)
```

Volumen y Area basal por familia

```python
#  KPI VOLUMEN Y ÁREA BASAL POR FAMILIA

from pyspark.sql.functions import sum, round, count, col

kpi_volumen_area_familia = (
    df_gold
    .groupBy("familia")
    .agg(
        count("*").alias("total_arboles"),
        round(sum("vol_total_m3"), 3).alias("volumen_total_m3"),
        round(sum("area_basal_m2"), 3).alias("area_basal_total_m2")
    )
    .orderBy(col("volumen_total_m3").desc())
)

display(kpi_volumen_area_familia)
```
